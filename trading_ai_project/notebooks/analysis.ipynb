# Trading Agent Analysis

```python
import os
import sys
sys.path.insert(0, os.path.abspath(os.path.join(os.path.dirname('__file__'), '..')))
from stable_baselines3 import PPO
from trading_ai_project.env.gym_env import TradingEnv
import pandas as pd
```

```python
db_path = os.path.join('..', 'database', 'trading_data.db')
env = TradingEnv(db_path)
```

```python
model_path = "../models/ppo_trading_agent.zip"
model = PPO.load(model_path, env=env)
```

```python
obs, info = env.reset()
done = False
actions = []
net_worths = []

while not done:
    action, _states = model.predict(obs, deterministic=True)
    obs, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated
    
    actions.append(action)
    net_worths.append(env.net_worth)
```

## Analyze Actions

```python
actions_df = pd.DataFrame(actions, columns=['action'])
action_counts = actions_df['action'].value_counts()
print(action_counts)
```

## Plot Net Worth

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(15, 5))
plt.plot(net_worths)
plt.title('Agent Net Worth Over Time')
plt.show()
```
